````python
import matplotlib.pyplot as plt
import numpy as np

# Swim times for two swimmers (in seconds)
swimmer1 = [39.01, 47.54, 46.86, 46.93, 50.96, 51.77, 43.66, 40.15]
swimmer2 = [37.38, 47.12, 47.73, 47.74, 49.58, 51.06, 43.75, 41.50]

# Calculate 100m splits (sum every two consecutive 50m splits)
splits1 = [swimmer1[i] + swimmer1[i+1] for i in range(0, len(swimmer1), 2)]
splits2 = [swimmer2[i] + swimmer2[i+1] for i in range(0, len(swimmer2), 2)]

def format_time(seconds):
    minutes = int(seconds // 60)
    secs = seconds % 60
    return f"{minutes}:{secs:05.2f}"

split_labels = ['Fly', 'Back', 'Breast', 'Free']

x = np.arange(len(split_labels))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
rects1 = ax.bar(x - width/2, splits1, width, label=f'Swimmer 1 (Total: {format_time(sum(swimmer1))})')
rects2 = ax.bar(x + width/2, splits2, width, label=f'Swimmer 2 (Total: {format_time(sum(swimmer2))})')

ax.set_ylabel('Time (s)')
ax.set_xlabel('Stroke')
ax.set_title('400m Medley 100m Splits Comparison (mm:ss.ss)')
ax.set_xticks(x)
ax.set_xticklabels(split_labels)
ax.legend(loc='lower right')

for i, rects in enumerate([rects1, rects2]):
    splits = splits1 if i == 0 else splits2
    for j, rect in enumerate(rects):
        height = rect.get_height()
        ax.annotate(format_time(splits[j]),
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom')

plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np

# Swim times for two swimmers (in seconds)
swimmer1 = [39.01, 47.54, 46.86, 46.93, 50.96, 51.77, 43.66, 40.15]
swimmer2 = [37.38, 47.12, 47.73, 47.74, 49.58, 51.06, 43.75, 41.50]

# Calculate 100m splits (sum every two consecutive 50m splits)
splits1 = [swimmer1[i] + swimmer1[i+1] for i in range(0, len(swimmer1), 2)]
splits2 = [swimmer2[i] + swimmer2[i+1] for i in range(0, len(swimmer2), 2)]

def format_time(seconds):
    minutes = int(seconds // 60)
    secs = seconds % 60
    return f"{minutes}:{secs:05.2f}"

split_labels = ['Fly', 'Back', 'Breast', 'Free']

# Prepare x-axis for 50m and 100m points
x_50 = np.arange(1, 9) * 50
x_100 = np.arange(1, 5) * 100

fig, ax = plt.subplots(figsize=(10, 6))

# Plot 50m splits
ax.plot(x_50, swimmer1, 'o-', label='Swimmer 1 50m', color='blue', alpha=0.7)
ax.plot(x_50, swimmer2, 'o-', label='Swimmer 2 50m', color='red', alpha=0.7)

# Plot 100m splits as larger circles
ax.scatter(x_100, splits1, s=200, c='blue', label='Swimmer 1 100m', edgecolors='black', zorder=3)
ax.scatter(x_100, splits2, s=200, c='red', label='Swimmer 2 100m', edgecolors='black', zorder=3)

# Annotate 100m splits with mm:ss.ss
for i, (x, s) in enumerate(zip(x_100, splits1)):
    ax.annotate(format_time(s), (x, splits1[i]), textcoords="offset points", xytext=(0,10), ha='center', color='blue')
for i, (x, s) in enumerate(zip(x_100, splits2)):
    ax.annotate(format_time(s), (x, splits2[i]), textcoords="offset points", xytext=(0,-15), ha='center', color='red')

ax.set_xticks(np.arange(0, 450, 50))
ax.set_xlabel('Distance (m)')
ax.set_ylabel('Time (s)')
ax.set_title('400m Medley: 50m and 100m Splits (Graph)')
ax.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````
